Partie Théorique + Exemples Simples

```python
from sympy import symbols, And, Or, Not, Implies
from sympy.logic.inference import satisfiable

# Définition des propositions
P, Q = symbols('P Q')

# Construction des expressions logiques
f1 = And(P, Q)
f2 = Or(P, Q)
f3 = Not(P)
f4 = Implies(P, Q)

# Vérification de la satisfiabilité
f5 = And(P, Not(P))

print("P ET Q :", f1)
print("P OU Q :", f2)
print("NON P :", f3)
print("P implique Q :", f4)
print("P ET NON P :", satisfiable(f5))
```

L'exécution du programme permet de représenter plusieurs opérations logiques avec SymPy. La fonction satisfiable() retourne False pour la formule P ET NON P, car cette expression est contradictoire.

```python

faits = {"P"}

regles = [
    ({"P"}, "Q"),
    ({"Q"}, "R")
]

# Chainage avant
changement = True

while changement:
    changement = False

    for premisses, conclusion in regles:
        if premisses.issubset(faits) and conclusion not in faits:
            faits.add(conclusion)
            changement = True
            print("Nouveau fait déduit :", conclusion)

print("Ensemble final des faits :", faits)
```

```python

faits = {"P"}


regles = [
    ({"P"}, "Q"),
    ({"Q"}, "R")
]

# Fonction de chainage arrière
def chainage_arriere(but):
    if but in faits:
        return True

    for premisses, conclusion in regles:
        if conclusion == but:
            if all(chainage_arriere(p) for p in premisses):
                return True

    return False


# Vérification de l'objectif
objectif = "R"

if chainage_arriere(objectif):
    print("L'objectif", objectif, "est démontré.")
else:
    print("L'objectif", objectif, "n'est pas démontré.")
```

```python

faits = {"P"}

regles = [
    ({"P"}, "Q"),
    ({"Q"}, "R")
]

# Fonction de fermeture
def fonction_fermeture(faits, regles):
    resultat = set(faits)
    changement = True

    while changement:
        changement = False

        for premisses, conclusion in regles:
            if premisses.issubset(resultat) and conclusion not in resultat:
                resultat.add(conclusion)
                changement = True

    return resultat


# Calcul de la fermeture
fermeture = fonction_fermeture(faits, regles)

print("Faits initiaux :", faits)
print("Fermeture :", fermeture)
```

Application pratique : Système expert basé sur des règles logiques

Présentation de l'exercice

Nous allons développer un système expert qui utilise des faits et des règles logiques pour déduire des conclusions concernant les problèmes possibles de démarrage d'un ordinateur.
Faits :
(P) : L'ordinateur ne démarre pas.
(Q) : Un problème d'alimentation est possible.
(R) : Il faut vérifier le chargeur.
(S) : La batterie est déchargée.

Règles d'inférence :
(P → Q)
(Q → R)
(S → Q)

Nous appliquerons ces règles pour étudier le chaînage avant, le chaînage arrière et la fonction de fermeture.

```python

faits = {
    "P"
}

regles = [
    ({"P"}, "Q"),
    ({"Q"}, "R"),
    ({"S"}, "Q")
]

print("Faits initiaux :", faits)
print("Règles d'inférence :", regles)
```

 Vérification logique des règles:
Nous allons utiliser `itertools` pour construire une table de vérité et `SymPy` pour représenter les règles logiques et vérifier si un objectif peut être déduit des faits disponibles.

```python
import itertools

print("P\tQ\tP -> Q")

for p, q in itertools.product([False, True], repeat=2):
    implication = (not p) or q
    print(f"{p}\t{q}\t{implication}")
```

Vérification logique avec SymPy

```python
from sympy import symbols, And, Not, Implies
from sympy.logic.inference import satisfiable


P_sym, Q_sym, R_sym, S_sym = symbols("P Q R S")

regles_sym = And(
    Implies(P_sym, Q_sym),
    Implies(Q_sym, R_sym),
    Implies(S_sym, Q_sym)
)

base_connaissances = And(P_sym, regles_sym)


contre_exemple = satisfiable(
    And(base_connaissances, Not(R_sym))
)

print("Règles logiques :", regles_sym)
print("L'objectif R est démontrable :", contre_exemple is False)
```

```python
def chainage_avant(faits, regles):
    resultat = set(faits)
    changement = True

    while changement:
        changement = False

        for premisses, conclusion in regles:
            if premisses.issubset(resultat):
                if conclusion not in resultat:
                    resultat.add(conclusion)
                    changement = True
                    print("Nouveau fait déduit :", conclusion)

    return resultat


faits_deduits = chainage_avant(faits, regles)

print("\nEnsemble final des faits :", faits_deduits)
```

```python
def chainage_arriere(but, faits, regles):

    if but in faits:
        return True

   
    for premisses, conclusion in regles:
        if conclusion == but:
            if all(chainage_arriere(p, faits, regles)
                   for p in premisses):
                return True

    return False

objectif = "R"

if chainage_arriere(objectif, faits, regles):
    print("L'objectif", objectif, "est démontré.")
else:
    print("L'objectif", objectif, "n'est pas démontré.")
```

```python
def fonction_fermeture(faits, regles):
    resultat = set(faits)
    changement = True

    while changement:
        changement = False

        for premisses, conclusion in regles:
            if premisses.issubset(resultat):
                if conclusion not in resultat:
                    resultat.add(conclusion)
                    changement = True

    return resultat


# Calcul de la fermeture
fermeture = fonction_fermeture(faits, regles)

print("Faits initiaux :", faits)
print("Fermeture :", fermeture)
```
